Function to take input of whether a person has a specific symptom or not and then return the value as {0} for - 'no' and {1} for 'yes'.

```python
def inp_yes_no(symptom) -> int:
    while True:
        yes_no = input(f"Enter 0 if you don't have {symptom}.\nEnter 1 if you have {symptom}.\n")
        if yes_no in ["0", "1"]:
            return int(yes_no)
        print("Invalid input. Please enter exactly 0 or 1.\n")
```

Function to return a dictionary of symptoms.

```python
def symp_list():
    d= {"Fever":0, "Cough":0,"Sore Throat":0, "Body Pain":0,"Headache":0}
    return d
```

Function taking in a dict of symptoms with values of {0} meaning 'no' and then returning the dict of symptoms with either values of 0 or 1.

```python
def symp_yes_no(list_of_symp):
    for key,value in list_of_symp.items():
        y_n=inp_yes_no(key)
        if(y_n==0):
            continue
        else:
            list_of_symp[key]=1
```

Function taking in the symptoms and returning(printing) the diagnosis baased on those symptoms.

```python
def healthy_or_infected(symptoms):
    if(symptoms["Fever"]==0 and symptoms["Cough"]==0 and symptoms["Sore Throat"]==0 and symptoms["Body Pain"]==0 and symptoms["Headache"]==0 ):
        print(f"You are Healthy.\n")
    elif(symptoms["Fever"]==1 and symptoms["Cough"]==1 and symptoms["Sore Throat"]==1 and symptoms["Body Pain"]==0 and symptoms["Headache"]==0 ):
        print(f"You have Flu.\n")
    elif(symptoms["Fever"]==1 and symptoms["Cough"]==0 and symptoms["Sore Throat"]==0 and symptoms["Body Pain"]==1 and symptoms["Headache"]==1 ):
        print(f"You have Dengue.\n")
    elif(symptoms["Fever"]==0 and symptoms["Cough"]==1 and symptoms["Sore Throat"]==1 and symptoms["Body Pain"]==0 and symptoms["Headache"]==0 ):
        print(f"You have Common Cold.\n")
    elif(symptoms["Fever"]==0 and symptoms["Cough"]==0 and symptoms["Sore Throat"]==0 and symptoms["Body Pain"]==0 and symptoms["Headache"]==1 ):
        print(f"You have Migrane.\n")
    else:
         print(f"Your conditioned cannot be listed.")
```

Main function to run the program.

```python
if __name__=="__main__":
    sl = symp_list()
    symp_yes_no(sl)
    healthy_or_infected(sl)
```
