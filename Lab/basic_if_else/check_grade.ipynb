Function to take number of subjects.

```python
def inp_total_subjects()->int:
    total_subs=int(input("Enter total number of subjects"))
    if(total_subs<=0):
        print(f"Subjects cannot be zero or negative.\n")
    return total_subs
```

Function to take individual marks of each subject.

```python
def inp_single_marks(i)->float:
    marks=float(input(f"Enter marks for subject {i+1}:\t"))
    return marks
```

Functions to take all marks of a student, subject wise.

```python
def inp_all_marks(total_subs)->list:
    marks_list=[]
    for i in range (total_subs):
        marks=inp_single_marks(i)
        if(marks<0 or marks>100):
            print(f"Marks cannot be less than zero or more than 100.\n")
            marks=inp_single_marks(i)
        marks_list.append(marks)
    return marks_list
```

Function to return average of all subjects.

```python
def marks_avg(all_marks)->float:
    return sum(all_marks)/len(all_marks)
```

Function to print the grades of the students based on some specific constraints.

```python
def grade_student(avg,attendance):
    if(avg>=75 and avg<=100):
        print(f"Your Grade is A.\n")
    elif (avg>=60 and avg<75):
        print(f"Your Grade is B.\n")
    elif(avg>=45 and avg<60):
        print(f"Your Grade is C.\n")
    elif(avg>=40 and avg<45):
        print(f"Your Grade is Pass Class.\n")
    elif(avg>=35 and avg<40):
        if(attendance>=75):
            avg+=5
            print(f"Your Grade is Pass Class.\n")
        else:
            print(f"You have Failed.\n")
    else:
        print(f"You have Failed.\n")
```

Function to take attendance of the student.

```python
def inp_attend():
    attend=float(input("Enter your attendance:\t"))
    if(attend<0 or attend >100):
        print(f"Attendance cannnot be negative or more than 100.\n")
        return inp_attend()
    return attend
```

Main functions to run the program.

```python
if __name__ == "__main__":
    attend=inp_attend()
    tot_subs=inp_total_subjects()
    all_marks=inp_all_marks(tot_subs)
    avg=marks_avg(all_marks)
    grade_student(avg, attend)
```
