```python
import heapq
```

```python
def a_star(graph, heuristic, start, goal):

    # Check if start and goal exist
    if start not in graph or goal not in graph:
        print("Invalid Start or Goal Node.")
        return

    # Priority Queue (Min Heap)
    priority_queue = []

    # Push starting node
    heapq.heappush(priority_queue, (heuristic[start], start))

    # Actual cost from start
    g_cost = {start: 0}

    # Parent dictionary for reconstructing path
    parent = {start: None}

    # Closed set (visited nodes)
    closed = set()

    while priority_queue:
        #we ignore the f_cost here since we only need the node for processing 
        # #but the priority queue is a list of tuples (f_cost, node) 
        # #so we unpack it accordingly but not use the f_cost variable
        _current_f_cost, current_node = heapq.heappop(priority_queue)

        # Skip if already processed
        if current_node in closed:
            continue

        closed.add(current_node)

        print(f"Visiting : {current_node}")

        # Goal reached
        if current_node == goal:

            path = []

            while current_node is not None:
                path.append(current_node)
                current_node = parent[current_node]

            path.reverse()

            print("\nShortest Path:")
            print(" -> ".join(path))

            print("Total Cost:", g_cost[goal])

            return

        # Explore neighbours
        for neighbour, weight in graph[current_node]:

            new_cost = g_cost[current_node] + weight

            if neighbour not in g_cost or new_cost < g_cost[neighbour]:

                g_cost[neighbour] = new_cost
                parent[neighbour] = current_node

                f_cost = new_cost + heuristic[neighbour]

                heapq.heappush(priority_queue, (f_cost, neighbour))

    print("No Path Found")
```

```python
if __name__ == "__main__":
    graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 6)],
    'E': [('G', 1)],
    'F': [('G', 2)],
    'G': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 2,
    'E': 1,
    'F': 2,
    'G': 0
}

start = input("Enter Start Node : ").upper()
goal = input("Enter Goal Node : ").upper()

a_star(graph, heuristic, start, goal)
```
