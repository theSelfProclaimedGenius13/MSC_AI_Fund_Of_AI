```python
from collections import deque
```

```python
def bfs(graph,start,target):
    visited=set()
    queue=deque([start])
    visited.add(start)
    while queue:
        vertex=queue.popleft()
        print(f"Node '{vertex}':")
        print(f"    \u2193")
        if vertex == target:
            print("Target found!")
            return
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
        
```
