```python
def dfs(graph,node,target,visited=None):
    if visited is None :
        visited=set()
    if node not in visited:
        print(f"Node '{node}':")
        print(f"    \u2193")
        visited.add(node)
        if node == target:
            return True
        for neighbour in graph[node]:
            if dfs(graph,neighbour,target,visited):
                return True
    
    return False
```
